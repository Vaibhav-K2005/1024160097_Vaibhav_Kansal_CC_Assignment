```python
#Assingment 6.1:
def add_odd(num):
  sum=0
  for i in range (1,num+1):
    if i % 2 != 0:
      sum+=i
  return sum

n=int(input("enter the number: "))
print(add_odd(n))
```

```python
#Assingment 6.2:
def is_prime(num):
    if num <2:
      return False
    for j in range (2,int(num**0.5)+1):
      if num%j==0:
        return False
    return True
n=int(input("enter the number: "))
sum=0
for i in range(2,n+1):
  if is_prime(i):
    sum+=i
print(sum)
```
