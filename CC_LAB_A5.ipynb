Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_vk=np.array([[10,20,30,40],
                   [50,60,70,80],
                   [90,15,20,35]])
print(ucs420_vk)
print("Mean:",np.mean(ucs420_vk))
print("Median:",np.median(ucs420_vk))
print("Max_val:",np.max(ucs420_vk))
print("Min_val:",np.min(ucs420_vk))
print("Unique_elements",np.unique(ucs420_vk))

reshaped_ucs420_vk=ucs420_vk.reshape(4,3)
print("reshaped:",reshaped_ucs420_vk);

resized_ucs420_vk=np.resize(ucs420_vk,(2,3))
print("resized:",resized_ucs420_vk)

```
