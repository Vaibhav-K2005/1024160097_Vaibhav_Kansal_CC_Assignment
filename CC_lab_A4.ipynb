Q2

```python
def score_query(query, df):
    results = []

    query_words = set(query.lower().split())

    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())

        matches = query_words & text_words

        if matches:
            confidence = len(matches) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    return pd.DataFrame(results).sort_values(
        "confidence", ascending=False
    )


# Test
query = "how can I pay my fee"

print(score_query(query, df))
```

```python
def score_query(query, df):
    results = []

    query_words = set(query.lower().split())

    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())

        matches = query_words & text_words

        if matches:
            confidence = len(matches) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })

    if not results:
        print("No matching entries.")
        return

    # Find highest score
    highest = max(x["confidence"] for x in results)

    # Keep ALL entries with highest score
    best_matches = [
        x for x in results
        if x["confidence"] == highest
    ]

    print("Best match(es):")
    for x in best_matches:
        print(x)


# -------------------------
# Query that produces a tie
# -------------------------
print("Query 1:")
score_query("fee", df)


# -------------------------
# Query that does not produce a tie
# -------------------------
print("\nQuery 2:")
score_query("password", df)
```
