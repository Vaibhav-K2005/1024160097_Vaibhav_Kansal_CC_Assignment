```python
roll_no = "1024160097"
L = [int(digit) * 10 for digit in roll_no]

print("Initial List:", L)

L.append(100)
print("After append(100):", L)
L.insert(2, 25)
print("After insert(2, 25):", L)
L.remove(25)
print("After remove(50):", L)

L.pop(3)
print("After pop(3):", L)
L.sort()
print("Ascending Order:", L)

L.sort(reverse=True)
print("Descending Order:", L)

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

average = sum(L) / len(L)

new_list = [x for x in L if x > average]

print("Elements greater than average:", new_list)
```

```python
scores = tuple(L[:8])
print("Scores tuple:", scores)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)

reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)

search_score = int(input("Enter a score to search: "))

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score not present in tuple")

first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)

scores[0] = 100
```

```python
import random
random.seed(1024160097)
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random Numbers:")
print(numbers)

odd_numbers = [x for x in numbers if x % 2 != 0]
print("\nOdd numbers:")
print(odd_numbers)
print("Number of odd numbers:", len(odd_numbers))

even_numbers = [x for x in numbers if x % 2 == 0]
print("\nEven numbers:")
print(even_numbers)
print("Number of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
prime_numbers = [x for x in numbers if is_prime(x)]
print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)
print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
roll_no = "1024160097"
digits = [int(d) for d in roll_no[:8]]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)

union_set = A.union(B)

print("\nUnion of A and B:", union_set)

intersection_set = A.intersection(B)

print("Intersection of A and B:", intersection_set)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

symmetric_diff = A.symmetric_difference(B)

print("Symmetric Difference:", symmetric_diff)
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))


X = int(input("\nEnter a value to remove from A: "))
A.discard(X)
print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Vaibhav",
    "roll_no": "1024160097",
    "branch": "COSE",
    "age": 20,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)

my_dict["cgpa"] = 9.3
print("\nAfter adding CGPA:")
print(my_dict)

my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

removed_branch = dict_pop.pop("branch")
print("\nAfter deleting branch using pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

del dict_del["branch"]
print("\nAfter deleting branch using del:")
print(dict_del)

print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")


print("\nChecking for email:")
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present.")

friend_dict = {
    "name": "Saket",
    "roll_no": "1024160028",
    "branch": "COSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nFriend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print("\nDictionary containing only string values:")
print(string_values)
```
